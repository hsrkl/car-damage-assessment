```python
from ultralytics import YOLO

# 1. Load the Segmentation model
model = YOLO('yolo26n-seg.pt')

# 2. Train with VALID parameters
results = model.train(
    data='/content/datasets/data.yaml',
    epochs=50,
    imgsz=640,
    batch=-1,
    device=0,
    name='yolo26_seg_runnin',

    # --- Valid Augmentations & Hyperparameters ---
    optimizer='MuSGD',     # Use 'auto' if this throws an error
    copy_paste=0.3,        # Valid: Randomly copies objects to new spots (Great for Seg)
    overlap_mask=True,     # Valid: Handles overlapping objects
    box=7.5,               # Valid: How much to prioritize Bounding Box accuracy
    cls=0.5,               # Valid: How much to prioritize Class accuracy
    patience=10            # Stop if no improvement after 10 epochs
)
```
